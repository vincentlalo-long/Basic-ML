```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_svmlight_file
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

### Sonar (Mine vs Rock) Dataset

The dataset describes the shape of sonars by comparing them to rocks or mines. Each object is a metal cylinder (mine) or a roughly cylindrical rock. There are 60 features (each between 0.0 and 1.0) and a target variable (M for mine, R for rock).

Let's load the dataset and prepare it for our perceptron.

```python
from sklearn.datasets import fetch_openml

# Load the Sonar dataset (ID: 40)
# The 'Sonar' dataset is available on OpenML with ID 40.
sonar = fetch_openml(name='sonar', version=1, as_frame=False, return_X_y=True)
X_sonar, y_sonar_str = sonar[0], sonar[1]
```

```python
print(X_sonar.shape)
print(y_sonar_str.shape)
print(X_sonar[:5])
print(y_sonar_str[:5])
print(np.unique(y_sonar_str))
```

```python
unique_labels, counts = np.unique(y_sonar_str, return_counts=True)

for label, count in zip(unique_labels, counts):
    print(f"{label}: {count}")
```

```python
# Convert target labels from 'M' (Mine) and 'R' (Rock) to 1 and -1
# 'M' will be 1, 'R' will be -1
y_sonar = np.where(y_sonar_str == 'M', 1, -1)
print(f"Features shape: {X_sonar.shape}")
print(f"Target shape: {y_sonar.shape}")
print("First 5 features:")
print(X_sonar[:5])
print("First 5 targets:")
print(y_sonar[:5])
```

```python
X_train , X_test, y_train, y_test = train_test_split(X_sonar, y_sonar, test_size=0.2, random_state=42, stratify=y_sonar)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"Scaled X_train shape: {X_train_scaled.shape}")
print(f"Scaled X_test shape: {X_test_scaled.shape}")
```

### Train the Perceptron Model

Now, let's train the `SimplePerceptron` with the scaled training data. We will use a smaller learning rate and a higher number of epochs to allow for better convergence, as the dataset is more complex than a simple 2D example.

```python
class SimplePerceptron:
    def __init__(self, learning_rate=0.1, epochs=100):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = None

    def fit(self, X, y, plot_during_training=False, plot_interval=1, X_plot=None, y_plot=None):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0

        for epoch in range(self.epochs):
            errors = 0
            for i in range(n_samples):
                linear_output = np.dot(X[i], self.weights) + self.bias
                # Activation function: step function (output -1 or 1)
                y_predicted = 1 if linear_output >= 0 else -1

                if y_predicted != y[i]:
                    # Update weights and bias
                    self.weights += self.learning_rate * y[i] * X[i]
                    self.bias += self.learning_rate * y[i]
                    errors += 1

            print(f"Epoch {epoch+1}/{self.epochs}, Errors: {errors}, Weights: {self.weights}, Bias: {self.bias}")

            if plot_during_training and (epoch + 1) % plot_interval == 0:
                if X_plot is not None and y_plot is not None:
                    plot_perceptron_decision_boundary(X_plot, y_plot, self, title=f"Epoch {epoch+1} Decision Boundary")
                else:
                    plot_perceptron_decision_boundary(X, y, self, title=f"Epoch {epoch+1} Decision Boundary")

        print(f"Final weights: {self.weights}, Final bias: {self.bias}")

    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        return np.where(linear_output >= 0, 1, -1)
```

```python
perceptron_model_scaled = SimplePerceptron(learning_rate=0.01, epochs=100)
perceptron_model_scaled.fit(X_train_scaled, y_train)
```

```python
# Evaluate model on train and test sets
y_train_pred = perceptron_model_scaled.predict(X_train_scaled)
y_test_pred = perceptron_model_scaled.predict(X_test_scaled)

train_accuracy = np.sum(y_train_pred == y_train) / len(y_train)
test_accuracy = np.sum(y_test_pred == y_test) / len(y_test)

print(f"Train Accuracy: {train_accuracy:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")

# Plot errors per epoch
plt.figure(figsize=(10, 5))
plt.plot(perceptron_model_scaled.errors_per_epoch, marker='o', linestyle='-')
plt.xlabel('Epoch')
plt.ylabel('Number of Errors')
plt.title('Training Errors per Epoch')
plt.grid(True)
plt.show()

print(f"\nEpoch khi errors = 0: {next((i+1 for i, e in enumerate(perceptron_model_scaled.errors_per_epoch) if e == 0), 'Never')}")
```

```python
def plot_perceptron_decision_boundary(X, y, model, title="Perceptron Decision Boundary"):
    plt.figure(figsize=(7, 5))
    x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
    y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5

    xx, yy = np.meshgrid(np.linspace(x_min, x_max, 100),
                         np.linspace(y_min, y_max, 100))

    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.coolwarm)

    # Plot data points
    plt.scatter(X[y == 1, 0], X[y == 1, 1], color='blue', marker='o', label='Class 1')
    plt.scatter(X[y == -1, 0], X[y == -1, 1], color='red', marker='x', label='Class -1')

    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
# Note: Cannot plot 2D decision boundary for 60-feature dataset
# Instead, let's plot predictions vs actual for a sample
print("Confusion Matrix (Train Set):")
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_train, y_train_pred)
print(cm)
print(f"\nConfusion Matrix (Test Set):")
cm_test = confusion_matrix(y_test, y_test_pred)
print(cm_test)
```
